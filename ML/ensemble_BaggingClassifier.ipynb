# Bagging Classifier

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.ensemble import BaggingClassifier

from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression

from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv(r'C:\Devesh ITV\Python\PythonProgramming\Datasets\IRIS.csv')
df.head()
```

```python
le = LabelEncoder()

df['species'] = le.fit_transform(df['species'])
df.head()
```

```python
# creating base models
from sklearn.neighbors import KNeighborsClassifier


dt = DecisionTreeClassifier()
svm = SVC(probability=True)
knn = KNeighborsClassifier()
```

```python
# train test split
X = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

```python
# bagging with DT

bc = BaggingClassifier(
    estimator=dt,
    bootstrap=True,
    bootstrap_features=False,
    n_estimators=100,
    max_samples=0.5,
    max_features=1.0,
    
)

cv = cross_val_score(bc, X, y, cv=10, scoring='accuracy')
print(bc.estimator, cv.mean())
```

```python
# bagging with SVM
bc = BaggingClassifier(
    estimator=svm,
    bootstrap=True,
    bootstrap_features=False,
    n_estimators=100,
    max_samples=0.5,
    max_features=1.0,
    
)

cv = cross_val_score(bc, X, y, cv=10, scoring='accuracy')
print(bc.estimator, cv.mean())
```

```python
# bagging with KNN

bc = BaggingClassifier(
    estimator=knn,
    bootstrap=True,
    bootstrap_features=False,
    n_estimators=100,
    max_samples=0.5,
    max_features=1.0,
    
)

cv = cross_val_score(bc, X, y, cv=10, scoring='accuracy')
print(bc.estimator, cv.mean())
```

```python
# Hyperparameter tuning for BaggingClassifier for DT
from sklearn.model_selection import GridSearchCV,RandomizedSearchCV

params = {
    'n_estimators': [10, 20, 50, 70, 100],
    'max_samples': [0.2, 0.5, 0.8, 1.0],
    'max_features': [0.2, 0.5, 0.8, 1.0],
    'bootstrap':[True,False],
    'bootstrap_features':[True,False]
}
```

```python
# using Decision Tree
grid = GridSearchCV(estimator=BaggingClassifier(estimator=dt),
                    param_grid=params,
                    cv=5,
                    scoring='accuracy',
                    n_jobs=-1)


grid.fit(X,y)
grid.best_params_, grid.best_score_
```

```python
# using SVM
grid = GridSearchCV(estimator=BaggingClassifier(estimator=svm),
                    param_grid=params,
                    cv=5,
                    scoring='accuracy',
                    n_jobs=-1,
                    verbose=0)


grid.fit(X,y)
grid.best_params_, grid.best_score_
```
