# Voting Classifier with Different algorithm

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Devesh ITV\Python\PythonProgramming\Datasets\IRIS.csv")
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
```

```python
df['species'] = le.fit_transform(df['species'])
```

```python
df.head()
```

```python
# selecting only 2 classes for binary classification
# selecting only 2 columns

new_df = df[df['species'] != 0][['sepal_length', 'sepal_width', 'species']]

```

```python
new_df.head()
```

```python
new_df.shape
```

```python
X = new_df.iloc[:,0:2]
y = new_df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.ensemble import VotingClassifier

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
```

```python
model1 = LogisticRegression()
model2 = KNeighborsClassifier()
model3 = SVC(probability=True)
```

```python
estimators = [('lr', model1), ('knn', model2), ('svc', model3)]
```

```python
# accuracy of each model using cross validation
for estimator in estimators:
    scores = cross_val_score(estimator[1], X, y, cv=5,scoring='accuracy')
    print(f"{estimator[0]}: {scores.mean()}")
```

# Hard voting

```python
vc = VotingClassifier(estimators=estimators, voting='hard')
scores = cross_val_score(vc, X, y, cv=5,scoring='accuracy')
print(scores.mean())
```

# Soft voting

```python
vc = VotingClassifier(estimators=estimators, voting='soft')
scores = cross_val_score(vc, X, y, cv=5,scoring='accuracy')
print(scores.mean())
```

# Voting Classifier with same algorithm

```python
# creating data with make_classificatiom
from sklearn.datasets import make_classification

X, y = make_classification(n_samples=1000, n_features=20, n_informative=15, n_redundant=5, random_state=2)
```

```python
svm1 = SVC(probability=True,degree=1,kernel='poly')
svm2 = SVC(probability=True,degree=2,kernel='poly')
svm3 = SVC(probability=True,degree=3,kernel='poly')
svm4 = SVC(probability=True,degree=4,kernel='poly')
svm5 = SVC(probability=True,degree=5,kernel='poly')
```

```python
estimators = [('svm1', svm1), ('svm2', svm2), ('svm3', svm3), ('svm4', svm4), ('svm5', svm5)]
```

```python
for estimator in estimators:
    scores = cross_val_score(estimator[1], X, y, cv=5,scoring='accuracy')
    print(f"{estimator[0]}: {scores.mean()}")
```

```python
vc1 = VotingClassifier(estimators=estimators, voting='hard')
scores = cross_val_score(vc1,X,y,cv=5,scoring='accuracy')
print(scores.mean())
```

```python
vc1 = VotingClassifier(estimators=estimators, voting='soft')
scores = cross_val_score(vc1,X,y,cv=5,scoring='accuracy')
print(scores.mean())
```
